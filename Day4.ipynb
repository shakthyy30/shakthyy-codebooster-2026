```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import year, month, to_date, col, round as spark_round
import matplotlib.pyplot as plt
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import year, month, to_date, col, round as spark_round
import matplotlib.pyplot as plt
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

#Create SparkSession

spark = SparkSession.builder \
    .appName('Day4_Bigdata_Sales') \
    .config('spark.sql.adaptive.enabled', 'true') \
    .getOrCreate()


print(f'Spark Version : {spark.version}')
print(f'SparkSession : ACTIVE')
print(f'Application : {spark.sparkContext.appName}')
```

```python
df_bronze = spark.read \
     .option('header', 'true') \
     .option('inferSchema', 'true') \
     .csv('large_sales_data.csv')

print('=== BRONZE LAYER - Raw Data ===')
print(f'Rows     : {df_bronze.count()}')
print(f'Names    : {df_bronze.columns}')
print()
df_bronze.printSchema()
```

```python
import os
def get_dir_size(path):
    """Get total size of a file or directory in KB."""
    if os.path.isfile(path):
        return os.path.getsize(path) / 1024
    total = 0
    for dirpath, dirnames, filenames in os.walk(path):
      for f in filenames:
        total += os.path.getsize(os.path.join(dirpath, f))
    return total / 1024

csv_size = get_dir_size('large_sales_data.csv')
parquet_size = get_dir_size('sales_bronze.parquet')
reduciton = (1 - (parquet_size / csv_size)) * 100

print(f"\nCSV size : {csv_size:.1f} KB")
print(f"Parquet size : {parquet_size:.1f} KB")
print(f'Reduction : {reduciton:.1f} % smaller')
print(f'\nAt 1 TB scale: CSV=1000 GB -> Parquet={1000*(1-(reduciton/100)):.1f} GB')
```

```python
df_bronze = df_bronze.withColumn('total_price_and_revenue', col('unit_price') + col('revenue'))
df_bronze.show(5, truncate=False)

df_bronze.printSchema()
```

```python
df_bronze.select("product","revenue","city").show()
```

```python
df_bronze.filter(F.col("revenue")>5000).show()
```

```python
df_bronze.groupBy("category") \
         .agg(F.sum("revenue").alias("total_revenue"),
              F.count("*").alias("orders"),
              F.avg("revenue").alias("avg_revenue")) \
         .orderBy("total_revenue", ascending=False)


```

```python
top_products = df_bronze \
    .groupBy("product") \
    .agg(
        F.sum('revenue').alias('total_revenue'),
        F.count('order_id').alias('num_orders'),
        F.avg('revenue').alias('avg_order_revenue')
    ) \
    .orderBy('total_revenue', ascending=False) \
    .limit(5)

print('=== Top 5 Products by Revenue ===')
top_products.show(truncate=False)
```

```python
# Aggregate data by region
revenue_by_region = df_bronze \
    .groupBy('region') \
    .agg(
        F.sum('revenue').alias('total_revenue'),
        F.count('order_id').alias('total_orders'),
        F.countDistinct('customer_name').alias('unique_customers')
    ) \
    .orderBy('total_revenue', ascending=False)

print('=== Revenue by Region ===')
revenue_by_region.show(truncate=False)
```

```python
monthly_revenue_trend = df_bronze \
    .withColumn('order_month_num', F.month(F.col('order_date'))) \
    .groupBy('order_month_num') \
    .agg(
        F.sum('revenue').alias('monthly_revenue'),
        F.count('order_id').alias('monthly_orders')
    ) \
    .orderBy('order_month_num') \
    .withColumnRenamed('order_month_num', 'order_month')

print('=== Monthly Revenue Trend ===')
monthly_revenue_trend.show(12)
```

```python
monthly_revenue_trend = df_bronze \
    .withColumn('order_month_num', F.month(F.col('order_date'))) \
    .groupBy('order_month_num') \
    .agg(
        F.sum('revenue').alias('monthly_revenue'),
        F.count('order_id').alias('monthly_orders')
    ) \
    .orderBy('order_month_num') \
    .withColumn('order_month', F.date_format(F.to_date(F.col('order_month_num').cast('string'), 'M'), 'MMMM')) \
    .drop('order_month_num') \
    .select('order_month', 'monthly_revenue', 'monthly_orders')

print('=== Monthly Revenue Trend ===')
monthly_revenue_trend.show(12)
```
