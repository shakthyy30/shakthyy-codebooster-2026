```python
def ask_llm(user_message, system_message="You are a helful assistant.", temperature=0.7, max_tokens=500):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_message},
            {"role": "user", "content": user_message},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content
test_response = ask_llm(
    "Where is sece? Answer in exactly 2 sentences."
)
print('=== LLM Response ===')
print(test_response)
```

```python
import json
import time


print("==================================================")
print("   RUNNING ALL PROMPTING TECHNIQUES TOGETHER      ")
print("==================================================\n")

# 1. ZERO-SHOT PROMPTING
print("--- [1] Zero-Shot Prompting ---")
zero_shot_address = "456 Brigade Road, Bangalore 560025, Karnataka, India"
zero_shot_response = ask_llm(
    user_message=f"Extract the city name from this address: {zero_shot_address}"
)
print(f"Zero-shot Result:\n{zero_shot_response}\n")

# 2. FEW-SHOT PROMPTING (With System Role Guardrails)

print("--- [2] Few-Shot Prompting (Structured JSON) ---")

json_system_message = (
    "You are a strict data parsing assistant. Your only task is to convert text to JSON. "
    "You must respond ONLY with a valid JSON object matching the requested format. "
    "Do not add markdown code blocks, introductory text, or explanations."
)

few_shot_prompt = """Convert employee text to JSON. Here are examples:

Input: RAMESH KUMAR, 45000, mumbai
Output: {"name": "RAMESH KUMAR", "salary": 45000, "city": "mumbai"}

Input: Meera, 85000, kolkata
Output: {"name": "Meera", "salary": 85000, "city": "kolkata"}

Input: ramesh kumar,45000,mumbai
Output:"""

few_shot_response = ask_llm(
    user_message=few_shot_prompt,
    system_message=json_system_message,
    temperature=0.0
)

print("Few-shot Raw Result:")
print(few_shot_response)

try:
    parsed = json.loads(few_shot_response.strip())
    print("\nSuccessfully parsed as JSON!")
    print(f"Parsed Data -> Name: {parsed['name']}, Salary: {parsed['salary']}, City: {parsed['city']}\n")
except json.JSONDecodeError:
    print("\nParsing failed - model added extra conversational text.")
    print("Solution: Tighten system constraints or strip backticks.\n")

# 3. ROLE-BASED PROMPTING & TEMPERATURE EXPERIMENT

print("--- [3] Role-Based Prompting & Temperature Experiment ---")

startup_system_message = (
    "You are an elite branding strategist and creative director for technology startups."
)
startup_prompt = "Give me one creative name for a data analytics startup."

for temp in [0.0, 0.5, 1.0]:
    response = ask_llm(
        user_message=startup_prompt,
        system_message=startup_system_message,
        temperature=temp
    )
    print(f"Temperature {temp} (Role: Branding Expert): {response.strip()}")
    time.sleep(1)

print("\nObservation:")
print("- temperature=0.0 -> Highly deterministic; identical or near-identical results across runs.")
print("- temperature=0.5 -> Balanced; introduces subtle vocabulary variations.")
print("- temperature=1.0 -> High creativity; ideal for brainstorming and divergent concepts.")
print("\nRule of Thumb: Use 0.0 for data pipelines/JSON parsing, and higher (0.7 - 1.0) for creative writing.")
```

```python
import json
import time


print("==========================================================")
print("   COMBINED PROMPTING METHODS (TOPIC: STARTUP NAMING)     ")
print("==========================================================\n")

# -----------------------------------------------------------------
# 1. ZERO-SHOT PROMPTING
# -----------------------------------------------------------------
print("--- [1] Zero-Shot Prompting ---")

zero_shot_prompt = "Give me one creative name for a data analytics startup."
zero_shot_response = ask_llm(user_message=zero_shot_prompt)

print("Zero-shot Result:")
print(zero_shot_response)
print("\n" + "-"*40 + "\n")


# -----------------------------------------------------------------
# 2. FEW-SHOT PROMPTING (With System Role Guardrails)
# -----------------------------------------------------------------
print("--- [2] Few-Shot Prompting (Structured JSON Outputs) ---")

json_system_message = (
    "You are a strict data parsing assistant. Your only task is to generate startup names in JSON. "
    "You must respond ONLY with a valid JSON object matching the requested format. "
    "Do not add markdown code blocks, conversational filler, or explanations."
)


few_shot_prompt = """Convert the tech domain into a structured JSON startup name. Here are examples:

Input: cloud computing
Output: {"company_name": "SkyCompute", "industry": "cloud", "uniqueness_score": 8.5}

Input: cybersecurity
Output: {"company_name": "AegisShield", "industry": "security", "uniqueness_score": 9.0}

Input: data analytics
Output:"""
few_shot_response = ask_llm(
    user_message=few_shot_prompt,
    system_message=json_system_message,
    temperature=0.0
)

print("Few-shot Raw Result:")
print(few_shot_response)

try:
    parsed = json.loads(few_shot_response.strip())
    print("\nSuccessfully parsed as JSON!")
    print(f"Parsed Data -> Name: {parsed['company_name']}, Industry: {parsed['industry']}, Score: {parsed['uniqueness_score']}\n")
except json.JSONDecodeError:
    print("\nParsing failed - model added conversational text.")
    print("Solution: Enforce strict system instructions.\n")

print("-"*40 + "\n")


# -----------------------------------------------------------------
# 3. ROLE-BASED PROMPTING & TEMPERATURE EXPERIMENT
# -----------------------------------------------------------------
print("--- [3] Role-Based Prompting & Temperature Experiment ---")
startup_system_message = (
    "You are an elite silicon valley branding strategist and creative director for technology startups."
)
startup_prompt = "Give me one creative name for a data analytics startup."

for temp in [0.0, 0.5, 1.0]:
    response = ask_llm(
        user_message=startup_prompt,
        system_message=startup_system_message,
        temperature=temp
    )
    print(f"Temperature {temp} (Role: Branding Expert): {response.strip()}")
    time.sleep(1)

print("\nObservation Summary:")
print("- temperature=0.0 -> Standard, reliable name choice (Highly deterministic).")
print("- temperature=0.5 -> Introduces slight variations and organic vocabulary shifts.")
print("- temperature=1.0 -> Highly experimental, abstract, and creative naming conventions.")
```
