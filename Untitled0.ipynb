```python
cs_students = df[df['department'] == 'Computer Science']


print(f"Number of Computer Science students:{len(cs_students)}")

print("\nComputer Science studentes:")
print(cs_students[['name','math_score','programming_score','attendance_percentage']])

```

```python
attendence = cs_students[cs_students['attendance_percentage'] > 90]
print("\nComputer Science students with attendence abpve 90%")
print(attendence[['name','math_score','programming_score','attendance_percentage']])
```

```python
mean_attendance = df['attendance_percentage'].mean()
print(f"Mean attendance percentage for all students: {mean_attendance:.2f}%")
```

```python
students_above_70_attendance = df[df['attendance_percentage'] > 70]
num_students_above_70 = len(students_above_70_attendance)
print(f"Number of students with attendance percentage above 70%: {num_students_above_70}")
print("\nStudents with attendance percentage above 70%:")
print(students_above_70_attendance[['name', 'department', 'attendance_percentage']])
```

```python
df['total_score'] = df['math_score'] + df['science_score'] + df['english_score'] + df['programming_score']
topper = df[df['total_score']/4 > 90]
print(topper[['name','total_score']])
print("\nNumber of topper: ", len(topper))
```

```python
df['average_score'] = (df['math_score'] + df['science_score'] + df['english_score'] + df['programming_score']) / 4
average_per_department = df.groupby('department')['average_score'].mean()
print(average_per_department)
```

```python
dept_avg = df.groupby('department')['total_score'].mean().round(2)
dept_avg_sort = dept_avg.sort_values(ascending=False)
print("Average Total Score ny Department:")
print(dept_avg_sort)
```
